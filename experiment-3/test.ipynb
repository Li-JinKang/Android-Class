```python
def selection_sort(arr):
    """
    选择排序算法实现
    参数:
        arr: 待排序的数值列表
    返回:
        排序后的列表
    """
    n = len(arr)
    # 遍历所有数组元素
    for i in range(n):
        # 找到从 i 到 n-1 中最小元素的索引
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
                
        # 将找到的最小元素与当前位置 i 的元素交换
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

def test():
    """
    选择排序算法测试函数
    """
    # 示例测试数据
    example_arr = [64, 25, 12, 22, 11]
    print("【测试示例】")
    print(f"排序前数组: {example_arr}")
    sorted_example = selection_sort(example_arr.copy())
    print(f"选择排序后: {sorted_example}")
    print("-" * 30)
    
    # 随机测试数据
    import random
    random_arr = [random.randint(1, 100) for _ in range(8)]
    print("【随机测试】")
    print(f"排序前随机数组: {random_arr}")
    sorted_random = selection_sort(random_arr.copy())
    print(f"选择排序后随机数组: {sorted_random}")

# 运行测试
test()
```

```python
# 1. 导入必要的库
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 设置绘图风格
sns.set(style="darkgrid")

# 支持中文显示
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'SimSun']  # 常用中文黑体
plt.rcParams['axes.unicode_minus'] = False  # 正常显示负号
```

```python
# 2. 读取并查看数据集
df = pd.read_csv('fortune500.csv')

# 统一重命名列名为小写，便于后续操作
df.columns = ['year', 'rank', 'company', 'revenue', 'profit']

print(f"数据集记录总数: {len(df)}")
print("\n数据集前 5 行:")
display(df.head())
print("\n数据列属性检查:")
df.info()
```

```python
# 3. 查找非数值利润值
non_numeric_profits = df.profit.str.contains('[^0-9.-]')
print(f"包含非数值利润的记录条数: {len(df.loc[non_numeric_profits])}")
print("\n非数值利润记录示例:")
display(df.loc[non_numeric_profits].head())
```

```python
# 4. 删除异常值行，并将利润列转换为数值型
df = df.loc[~non_numeric_profits]
df.profit = df.profit.apply(pd.to_numeric)

print(f"清洗后的数据集记录总数: {len(df)}")
print("\n清洗后的数据列属性:")
df.info()
```

```python
# 1. 按年份分组，计算每年的收入和利润均值
group_by_year = df.loc[:, ['year', 'revenue', 'profit']].groupby('year')
avgs = group_by_year.mean()

# 2. 创建画布与双 Y 轴绘制
fig, ax1 = plt.subplots(figsize=(12, 6), dpi=100)

# 绘制收入曲线（左轴）
color = 'tab:blue'
ax1.set_xlabel('Year (年份)', fontsize=12, labelpad=10)
ax1.set_ylabel('Mean Revenue (in millions) / 平均收入（百万美元）', color=color, fontsize=12)
line1 = ax1.plot(avgs.index, avgs.revenue, color=color, linewidth=2, label='平均收入 (Revenue)')
ax1.tick_params(axis='y', labelcolor=color)
ax1.grid(True, linestyle=':', alpha=0.6)

# 创建共享 X 轴的右轴
ax2 = ax1.twinx()  
color = 'tab:red'
ax2.set_ylabel('Mean Profit (in millions) / 平均利润（百万美元）', color=color, fontsize=12)
line2 = ax2.plot(avgs.index, avgs.profit, color=color, linewidth=2, linestyle='--', label='平均利润 (Profit)')
ax2.tick_params(axis='y', labelcolor=color)

# 合并图例
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left', fontsize=11)

# 设置标题
plt.title('Fortune 500 Mean Revenue and Profit Trends (1955-2005)\n财富500强企业平均收入与利润趋势图（双Y轴）', fontsize=14, pad=15)

# 自适应布局并展示
fig.tight_layout()
plt.show()
```
